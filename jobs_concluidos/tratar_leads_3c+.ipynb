```python
import pandas as pd
import re
import os
```

```python
df = pd.read_csv('data/bf25_inscritos/bf25_leads_inscritos.csv')
df_a = pd.read_csv('data/bf25_inscritos/bf25_alunos_inscritos.csv')

names = pd.read_csv('data/elementor_sub.csv')
```

```python
"""
Cell generated by Data Wrangler.
"""
def clean_data(df):
    # Drop duplicate rows in column: 'lead_id'
    df = df.drop_duplicates(subset=['lead_id'])
    # Drop rows with missing data in column: 'formatted_phone'
    df = df.dropna(subset=['formatted_phone'])
    # Drop duplicate rows in column: 'formatted_phone'
    df = df.drop_duplicates(subset=['formatted_phone'])
    return df

df = clean_data(df_a.copy())
```

```python
# Preenche os valores ausentes em 'name' no df com valores correspondentes do df names via 'email'
# Faz um merge para cruzar informações e preencher apenas onde falta
df = df.merge(
    names[['email', 'name']].rename(columns={'name': 'name_from_names'}),
    on='email',
    how='left'
)
df['name'] = df['name'].combine_first(df['name_from_names'])
df = df.drop(columns=['name_from_names'])
```

```python
alunos_tmb = pd.read_csv('data/alunos/alunos_tmb.csv',sep=';')
alunos_hotmart = pd.read_csv('data/alunos/alunos_hotmart.csv')
alunos_generic = pd.read_csv('data/alunos/alunos_generic.csv')

def clean_data(alunos_tmb):
    # Drop columns: 'Pedido', 'Produtor' and 31 other columns
    alunos_tmb = alunos_tmb.drop(columns=['Pedido', 'Produtor', 'Produto', 'Cliente CPF', 'Endereço completo', 'Logradouro', 'Número', 'Bairro', 'Complemento', 'CEP', 'Cidade', 'Estado', 'País', 'Status', 'Status Financeiro', 'Status Cancelamento', 'Ticket (R$)', 'Modalidade de Contrato', 'Criado Em', 'Data Efetivado', 'Data Cancelado', 'utm_source', 'utm_medium', 'utm_campaign', 'utm_content', 'utm_last_source', 'utm_last_medium', 'utm_last_campaign', 'utm_last_content', 'Renovação automatica', 'Data de Renovação Prevista', 'Renovação enviada', 'Oferta'])
    # Drop column: 'Cliente Nome'
    alunos_tmb = alunos_tmb.drop(columns=['Cliente Nome'])
    # Rename column 'Cliente Email' to 'email'
    alunos_tmb = alunos_tmb.rename(columns={'Cliente Email': 'email'})
    # Rename column 'Telefone' to 'phone'
    alunos_tmb = alunos_tmb.rename(columns={'Telefone': 'phone'})
    # Convert text to lowercase in column: 'email'
    alunos_tmb['email'] = alunos_tmb['email'].str.lower()
    # Remove leading and trailing whitespace in columns: 'email', 'phone'
    alunos_tmb['email'] = alunos_tmb['email'].str.strip()
    alunos_tmb['phone'] = alunos_tmb['phone'].str.strip()
    def sanitize_phone(x):
        sanitized = re.sub(r'[^0-9]', '', str(x))
        return sanitized if sanitized else pd.NA
    alunos_tmb['new_phone'] = alunos_tmb['phone'].apply(sanitize_phone)
    return alunos_tmb

alunos_tmb = clean_data(alunos_tmb.copy())


def clean_data(alunos_generic):
    # Convert text to lowercase in column: 'email'
    alunos_generic['email'] = alunos_generic['email'].str.lower()
    # Remove leading and trailing whitespace in columns: 'email', 'phone'
    alunos_generic['email'] = alunos_generic['email'].str.strip()
    alunos_generic['phone'] = alunos_generic['phone'].str.strip()
    def sanitize_phone(x):
        sanitized = re.sub(r'[^0-9]', '', str(x))
        return sanitized if sanitized else pd.NA
    alunos_generic['phone'] = alunos_generic['phone'].apply(sanitize_phone)
    return alunos_generic

alunos_generic = clean_data(alunos_generic.copy())

emails_to_exclude = set()
emails_to_exclude.update(alunos_tmb.email.unique())
emails_to_exclude.update(alunos_hotmart.email.unique())
emails_to_exclude.update(alunos_generic.email.unique())

phones_to_exclude = set()
phones_to_exclude.update(alunos_tmb.phone.unique())
phones_to_exclude.update(alunos_generic.phone.unique())
phones_to_exclude.update(df_3c.number.unique())
```

```python
"""
Cell generated by Data Wrangler.
"""
def clean_data(df):
    # Drop columns: 'biggest_fluency_challange', 'renda' and 2 other columns
    df = df.drop(columns=['biggest_fluency_challange', 'renda', 'utm_source', 'score'])

    df = df[~df['lead_id'].isin(df_a['lead_id'].unique())]
    df = df[~df['lead_id'].isin(df_3c.identifier.unique())]
    df = df[~df['email'].isin(emails_to_exclude)]
    df = df[~df['formatted_phone'].isin(phones_to_exclude)]

    df['type'] = df['formatted_phone'].apply(
        lambda x: 'Nacional' if str(x).startswith('+55') else 'Internacional'
    )
    
    # Filter rows based on column: 'type'
    df = df[df['type'] == "Nacional"]
    # Replace missing values with "Não informado" in columns: 'principal_objetivo', 'ja_fez_outro_curso', 'faixa_de_idade'
    df = df.fillna({'principal_objetivo': "Não informado", 'ja_fez_outro_curso': "Não informado", 'faixa_de_idade': "Não informado"})

    # Replace all instances of "+55" with "" in column: 'formatted_phone'
    df['formatted_phone'] = df['formatted_phone'].str.replace("+55", "", case=False, regex=False)

    df = df.drop(columns=['origem', 'faixa_de_idade', 'ja_fez_outro_curso', 'principal_objetivo', 'qualificado_geral', 'type'])
    
    df = df.drop_duplicates(subset=['formatted_phone'])
    df = df.drop_duplicates(subset=['lead_id'])
    return df

bf_base = clean_data(df.copy())
bf_base = bf_base.drop(columns=['renda_qualificada'])
```

```python
bf_base_rq = bf_base[bf_base['renda_qualificada'] == True]
bf_base_rd = bf_base[bf_base['renda_qualificada'] == False]

bf_base_rq = bf_base_rq.drop(columns=['renda_qualificada'])
bf_base_rd = bf_base_rd.drop(columns=['renda_qualificada'])


bf_base_rq.to_csv('export/lig_renda_qualificada.csv', index = False)
bf_base_rd.to_csv('export/lig_renda_desqualificada.csv', index = False)
```
