```python
df = pd.read_csv('data/base_alunos.csv')

"""
Cell generated by Data Wrangler.
"""
def clean_data(df):
    # Drop duplicate rows in column: 'phone'
    df = df.drop_duplicates(subset=['phone'])
    # Drop rows with missing data in column: 'phone'
    df = df.dropna(subset=['phone'])
    df = df.astype({'phone': 'string'})
    df['phone'] = df['phone'].apply(lambda x: x.replace('.0', ''))
    return df

df = clean_data(df.copy())
```

```python
alunos_prospects = pd.read_csv('data/alunos_prospects.csv', sep=';')

def clean_data(alunos_prospects):
    # Drop columns: 'Nome do Produtor', 'Documento do Produtor' and 52 other columns
    alunos_prospects = alunos_prospects.drop(columns=['Nome do Produtor', 'Documento do Produtor', 'Nome do Afiliado', 'Transação', 'Meio de Pagamento', 'Origem', 'Moeda', 'Preço do Produto', 'Moeda.1', 'Preço da Oferta', 'Taxa de Câmbio', 'Preço Original', 'Moeda.2', 'Número da Parcela',  'Data de Venda', 'Data de Confirmação', 'Status', 'Documento', 'CEP', 'Cidade', 'Estado', 'Bairro', 'País', 'Endereço', 'Número', 'Complemento', 'chave', 'Código do Produto', 'Código da Afiliação', 'Código de Oferta', 'Origem de Checkout', 'Tipo de Pagamento', 'Período Grátis', 'Tem co-produção', 'Venda feita como', 'Preço Total', 'Tipo pagamento oferta', 'Taxa de Câmbio Real', 'Preço Total Convertido', 'Quantidade de itens', 'Oferta de Upgrade', 'Cupom', 'Moeda.3', 'Valor que você recebeu convertido', 'Taxa de Câmbio do valor recebido', 'Data Vencimento', 'Instagram', 'Origem da venda', 'Moeda de recebimento', 'Código do assinante', 'Faturamento líquido', 'Nota Fiscal', 'Valor do frete bruto'])
    # Rename column 'Nome do Produto' to 'produtct'
    alunos_prospects = alunos_prospects.rename(columns={'Nome do Produto': 'produtct'})
    # Rename column 'Nome' to 'name'
    alunos_prospects = alunos_prospects.rename(columns={'Nome': 'name'})
    # Rename column 'Email' to 'email'
    alunos_prospects = alunos_prospects.rename(columns={'Email': 'email'})
    # Rename column 'DDD' to 'ddd'
    alunos_prospects = alunos_prospects.rename(columns={'DDD': 'ddd'})
    # Rename column 'Telefone' to 'raw_phone'
    alunos_prospects = alunos_prospects.rename(columns={'Telefone': 'raw_phone'})
    # Change column type to string for column: 'ddd'
    alunos_prospects = alunos_prospects.astype({'ddd': 'string'})
    # Round column 'raw_phone' (Number of decimals: 0)
    alunos_prospects = alunos_prospects.round({'raw_phone': 0})
    # Change column type to string for column: 'raw_phone'
    alunos_prospects = alunos_prospects.astype({'raw_phone': 'string'})
    alunos_prospects['ddd'] = alunos_prospects['ddd'].apply(lambda x: x.replace('.0', '') if pd.notna(x) else x)
    alunos_prospects['raw_phone'] = alunos_prospects['raw_phone'].apply(lambda x: x.replace('.0', '') if pd.notna(x) else x)
    # Drop rows with missing data in column: 'raw_phone'
    alunos_prospects = alunos_prospects.dropna(subset=['raw_phone'])
    # Filter rows based on column: 'Recorrência'
    alunos_prospects = alunos_prospects[(alunos_prospects['Recorrência'] == 0) | (alunos_prospects['Recorrência'] == 1) | (alunos_prospects['Recorrência'].isna())]
    # Drop column: 'Recorrência'
    alunos_prospects = alunos_prospects.drop(columns=['Recorrência'])
    # Filter rows based on column: 'produtct'
    alunos_prospects = alunos_prospects[alunos_prospects['produtct'] == "Cronograma dos Fluentes "]
    alunos_prospects['phone'] = alunos_prospects.apply(lambda x: f"55{x['ddd']}{x['raw_phone']}" if pd.notna(x['ddd']) else x['raw_phone'], axis=1)
    # Drop columns: 'ddd', 'raw_phone'
    alunos_prospects = alunos_prospects.drop(columns=['ddd', 'raw_phone'])
    return alunos_prospects

alunos_prospects = clean_data(alunos_prospects.copy())
```

```python
all_students = pd.concat([df,alunos_prospects])
```
